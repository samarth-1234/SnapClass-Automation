```python
import numpy as np
from sklearn.datasets import load_iris
iris = load_iris()
x=iris.data
y=iris.target
data = np.column_stack((x,y))
np.random.seed(42)
np.random.shuffle(data)

k=3
fold_size = len(data)//k
accuracies = []

for fold in range(k):

  start = fold * fold_size
  end = start + fold_size
  test_data = data[start:end]
  train_data = np.concatenate((data[:start], data[end:]))
  x_train, y_train = train_data[:, :-1], train_data[:, -1].astype(int)
  x_test, y_test = test_data[:, :-1], test_data[: , -1].astype(int)
  from collections import Counter
  most_common_class = Counter(y_train).most_common(1)[0][0]
  y_pred = np.full_like(y_test, fill_value = most_common_class)

  accuracy = np.mean(y_pred == y_test)
  accuracies.append(accuracy)
  print(f"Fold {fold + 1}:Accuracy = {accuracy:.2f}")

print(f"\n Average Accuracy over {k} folds : {np.mean(accuracies):.4f}")
```
